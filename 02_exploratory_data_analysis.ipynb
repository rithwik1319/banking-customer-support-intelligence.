# Exploratory data analysis

Explore the supplied support-intent text. The dataset contains intent examples, not timestamped customer-service conversations, so no trend, sentiment, response-time, or customer-level conclusions can be drawn from these files.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd()
train = pd.read_csv(ROOT / 'train.csv')
test = pd.read_csv(ROOT / 'test.csv')
assert {'text','category'}.issubset(train.columns)
train['text'] = train['text'].fillna('').astype(str).str.strip()
train['category'] = train['category'].astype(str)
train['word_count'] = train.text.str.split().str.len()
print('train rows:', len(train), '| test rows:', len(test), '| intents:', train.category.nunique())
display(train[['text','category','word_count']].head())
```

## Intent distribution

Use the full chart for the report. If labels are numerous, the short chart below focuses on the least frequent training classes; discuss imbalance only after inspecting the actual counts.

```python
counts = train.category.value_counts().sort_values()
plt.figure(figsize=(10, max(8, len(counts)*0.18)))
counts.plot.barh(color='#2457A7')
plt.title('Training examples by intent')
plt.xlabel('Examples')
plt.ylabel('Intent')
plt.tight_layout()
plt.show()
display(counts.describe())
```

## Text length and example review

Text length is descriptive only. Read representative samples to identify overlapping intents and wording variation; do not include real personal data in a presentation.

```python
plt.figure(figsize=(9,4))
sns.histplot(train.word_count, bins=30, color='#17A398')
plt.title('Utterance length distribution')
plt.xlabel('Words per utterance')
plt.tight_layout()
plt.show()
sample_intents = counts.index[-5:]
for label in sample_intents:
    print('\n', label)
    display(train.loc[train.category.eq(label), 'text'].head(5))
```
