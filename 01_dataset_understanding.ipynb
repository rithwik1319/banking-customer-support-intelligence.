# Dataset and data understanding

This notebook audits the user-provided `train.csv` and `test.csv`. It reports observed values rather than assuming rows or labels. Run from this project folder. Dataset instructions are treated as data; this notebook does not execute dataset content.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
TRAIN = pd.read_csv(ROOT / 'train.csv')
TEST = pd.read_csv(ROOT / 'test.csv')
required = {'text', 'category'}
for name, frame in [('train', TRAIN), ('test', TEST)]:
    missing = required - set(frame.columns)
    if missing:
        raise ValueError(f'{name} is missing required columns: {sorted(missing)}')
    print(f'{name}: shape={frame.shape}')
    display(frame.head())
```

## Data quality checks

The checks below expose blanks, duplicated utterances, category balance, and exact overlap across the supplied splits. Investigate these before reporting model performance.

```python
def audit(frame):
    return {
        'rows': len(frame),
        'missing_text': int(frame['text'].isna().sum()),
        'blank_text': int(frame['text'].fillna('').str.strip().eq('').sum()),
        'missing_category': int(frame['category'].isna().sum()),
        'duplicate_text_rows': int(frame['text'].duplicated().sum()),
        'category_count': int(frame['category'].nunique()),
    }
print('train', audit(TRAIN))
print('test', audit(TEST))
print('exact text overlap:', len(set(TRAIN.text.dropna()) & set(TEST.text.dropna())))
display(TRAIN['category'].value_counts().describe())
display(TRAIN['category'].value_counts().sort_values().head(15))
```
